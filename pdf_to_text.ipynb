```python
import os
import PyPDF2
import json

def extract_text_from_pdf(path_pdf):
    text = ""
    with open(path_pdf, "rb") as pdf_file:
        reader = PyPDF2.PdfReader(pdf_file)
        for page_num in range(len(reader.pages)):
            page = reader.pages[page_num]
            text += page.extract_text()
    return text

def split_text_into_chunks(text, chunk_size=256):
    words = text.split()
    all_chunks = [" ".join(words[i:i + chunk_size]) for i in range(0, len(words), chunk_size)]
    return all_chunks
```

```python
pdf_directory = "star_wars_pdfs/"
output_file_path = os.path.join(pdf_directory, "all_texts.jsonl")
with open(output_file_path, "w", encoding="utf-8") as output_file:
    for pdf_file_name in os.listdir(pdf_directory):
        if pdf_file_name.endswith(".pdf"):
            pdf_path = os.path.join(pdf_directory, pdf_file_name)
            book_text = extract_text_from_pdf(pdf_path)
            chunks = split_text_into_chunks(book_text)
            for chunk in chunks:
                json.dump({"text": chunk}, output_file, ensure_ascii=False)
                output_file.write("\n")
```
