```python
from elasticsearch import Elasticsearch
from dotenv import load_dotenv
import os

load_dotenv()

es = Elasticsearch(os.getenv("ES_HOST"), basic_auth=(os.getenv("ES_USER"), os.getenv("ES_PASSWORD")))

# read data from transcripts_headlines_joined.jsonl
with open("transcripts_headlines_joined.jsonl", "r") as f:
    for line in f:
        es.index(index="transcripts", body=line)
```
